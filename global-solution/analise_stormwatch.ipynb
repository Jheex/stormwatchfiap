# **1. Apresentação do Projeto StormWatch**

<a id="sec-1-1"></a>
## **1.1 Sobre o projeto**

O **StormWatch** é um projeto voltado ao monitoramento e à análise de dados ambientais, com foco em informações pluviométricas, hidrológicas e geotécnicas.

A proposta é reunir diferentes bases de dados ambientais em uma plataforma que permita organizar, visualizar e analisar informações relacionadas às condições monitoradas.

Por meio da utilização de técnicas de Ciência de Dados, o projeto busca identificar padrões, variações e possíveis anomalias nos registros, contribuindo para uma compreensão mais ampla dos fenômenos ambientais analisados.

<a id="sec-1-2"></a>
## **1.2 Problema**

O monitoramento ambiental envolve diferentes tipos de informações, que podem estar distribuídas em bases de dados distintas. Essa fragmentação dificulta a consulta conjunta, a comparação e a interpretação dos registros.

Além disso, inconsistências, valores ausentes e diferenças na estrutura dos dados podem comprometer as análises quando não são devidamente tratados.

Diante desse cenário, o StormWatch propõe organizar e integrar informações ambientais para facilitar sua consulta e análise.

Desenvolver uma solução para organização, integração, visualização e análise de dados pluviométricos, hidrológicos e geotécnicos, utilizando técnicas de Ciência de Dados para identificar padrões e variações nas informações ambientais.

<a id="sec-1-4"></a>
## **1.4 Objetivos específicos**

- Organizar diferentes bases de dados ambientais.
- Realizar a limpeza e a padronização das informações.

# **3. Limpeza e preparação dos dados**


<a id="sec-3-1"></a>
## **3.1 Estado inicial dos dados**

Após a importação dos três arquivos CSV, foi realizada uma verificação inicial para compreender a estrutura e a qualidade das bases antes de iniciar as análises.

As bases apresentaram os seguintes volumes:

- **Dados pluviométricos:** 468.164 registros;
- **Dados hidrológicos:** 25.072 registros;
- **Dados geotécnicos:** 244.449 registros.

**Total: 737.685 registros.**

Inicialmente, foram verificadas as colunas disponíveis, os tipos de dados, os períodos das medições, os valores ausentes, os possíveis registros duplicados, as categorias de sensores e a validade das coordenadas geográficas.

Essa verificação permitiu identificar situações que precisavam de tratamento antes da utilização dos dados nas análises estatísticas.

````python
# ============================================================
# 3.1 - ESTADO INICIAL DOS DADOS
# ============================================================

# ============================================================
# IMPORTAÇÃO DAS BASES DE DADOS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Definir os caminhos dos arquivos CSV
caminho_pluviometrico = "dados/dados_pluviometricos.csv"
caminho_hidrologico = "dados/dados_hidrologicos.csv"
caminho_geotecnico = "dados/dados_geotecnicos.csv"

# Carregar as três bases
pluviometricos = pd.read_csv(
    caminho_pluviometrico,
    sep=None,
    engine="python",
    encoding="utf-8-sig"
)

hidrologicos = pd.read_csv(
    caminho_hidrologico,
    sep=None,
    engine="python",
    encoding="utf-8-sig"
)

geotecnicos = pd.read_csv(
    caminho_geotecnico,
    sep=None,
    engine="python",
    encoding="utf-8-sig"
)

print("IMPORTAÇÃO DAS TRÊS BASES CONCLUÍDA.")
print(f"Pluviométricos: {len(pluviometricos):,}")
print(f"Hidrológicos: {len(hidrologicos):,}")
print(f"Geotécnicos: {len(geotecnicos):,}")

# Organizar as três bases para facilitar a inspeção
bases = {
    "Pluviométricos": pluviometricos,
    "Hidrológicos": hidrologicos,
    "Geotécnicos": geotecnicos
}

print("=" * 70)
print("STORMWATCH - DIAGNÓSTICO INICIAL DAS BASES DE DADOS")
print("=" * 70)

total_registros = 0

# Inspecionar cada base individualmente
for nome, df in bases.items():

    print("\n" + "=" * 70)
    print(f"BASE: {nome.upper()}")
    print("=" * 70)

    # Quantidade de registros e colunas
    print(f"Quantidade de registros: {df.shape[0]:,}")
    print(f"Quantidade de colunas: {df.shape[1]}")

    total_registros += df.shape[0]

    # Verificar colunas e tipos de dados
    print("\nCOLUNAS E TIPOS DE DADOS:")
    print(df.dtypes)

    # Verificar valores ausentes
    print("\nVALORES AUSENTES POR COLUNA:")
    print(df.isnull().sum())

    # Verificar possíveis registros duplicados
    print("\nQUANTIDADE DE REGISTROS DUPLICADOS:")
    print(df.duplicated().sum())

    # Verificar período das medições
    if "datahora" in df.columns:
        datas = pd.to_datetime(df["datahora"], errors="coerce")

        print("\nPERÍODO DAS MEDIÇÕES:")
        print(f"Data inicial: {datas.min()}")
        print(f"Data final: {datas.max()}")

    # Verificar categorias de sensores
    if "sensor" in df.columns:
        print("\nCATEGORIAS DE SENSORES:")
        print(df["sensor"].value_counts(dropna=False))

    # Verificar coordenadas geográficas
    if "latitude" in df.columns and "longitude" in df.columns:

        print("\nVERIFICAÇÃO DAS COORDENADAS:")

        latitude = pd.to_numeric(df["latitude"], errors="coerce")
        longitude = pd.to_numeric(df["longitude"], errors="coerce")

        coordenadas_invalidas = (
            ~latitude.between(-90, 90)
            | ~longitude.between(-180, 180)
        )

        print(
            "Registros com coordenadas fora dos limites válidos:",
            coordenadas_invalidas.sum()
        )

# Apresentar o total geral
print("\n" + "=" * 70)
print("RESUMO GERAL")
print("=" * 70)

print(f"Total de registros nas três bases: {total_registros:,}")

print("\nDIAGNÓSTICO INICIAL CONCLUÍDO.")
````

<a id="sec-3-2"></a>
## **3.2 Padronização e organização dos dados**

Foram realizados procedimentos para preparar as informações e permitir sua utilização conjunta nas análises.

Os principais procedimentos foram:

- **Datas e horários:** conversão e padronização da coluna `datahora` para permitir a análise temporal dos registros;
- **Valores numéricos:** verificação e conversão da coluna `valorMedida` para possibilitar cálculos estatísticos;
- **Categorias de sensores:** identificação e conferência dos diferentes tipos de sensores presentes nas bases;
- **Organização das bases:** separação dos dados pluviométricos, hidrológicos e geotécnicos em estruturas próprias para análise.

Esses procedimentos foram necessários para garantir que as informações apresentassem formatos adequados para os cálculos e comparações posteriores.

````python

# ============================================================
# 3.2 - PADRONIZAÇÃO E ORGANIZAÇÃO DOS DADOS
# ============================================================

# Criar cópias para preservar as bases originais
pluviometricos_limpa = pluviometricos.copy()
hidrologicos_limpa = hidrologicos.copy()
geotecnicos_limpa = geotecnicos.copy()

# Organizar as bases para aplicar os procedimentos
bases_limpeza = {
    "Pluviométricos": pluviometricos_limpa,
    "Hidrológicos": hidrologicos_limpa,
    "Geotécnicos": geotecnicos_limpa
}

print("=" * 70)
print("STORMWATCH - PADRONIZAÇÃO E ORGANIZAÇÃO DOS DADOS")
print("=" * 70)

for nome, df in bases_limpeza.items():

    print(f"\nBASE: {nome.upper()}")

    # Padronizar a coluna de data e horário
    if "datahora" in df.columns:
        df["datahora"] = pd.to_datetime(
            df["datahora"],
            errors="coerce"
        )

        print("Coluna datahora convertida para formato datetime.")

    # Converter os valores de medição para formato numérico
    if "valorMedida" in df.columns:
        df["valorMedida"] = pd.to_numeric(
            df["valorMedida"],
            errors="coerce"
        )

        print("Coluna valorMedida convertida para formato numérico.")

    # Padronizar as categorias dos sensores
    if "sensor" in df.columns:
        df["sensor"] = (
            df["sensor"]
            .astype("string")
            .str.strip()
            .str.lower()
        )

        print("Categorias de sensores padronizadas.")

    print(f"Registros disponíveis: {len(df):,}")

print("\nPADRONIZAÇÃO E ORGANIZAÇÃO CONCLUÍDAS.")
````

<a id="sec-3-3"></a>
## **3.3 Verificação de registros duplicados**

Foi realizada uma verificação de possíveis registros duplicados utilizando a identificação da estação e a data e hora da medição. Nas bases hidrológica e geotécnica, também foi considerado o tipo de sensor.

A verificação não identificou duplicidades nas combinações de campos utilizadas.

Por esse motivo, não foi necessária a remoção de registros por duplicidade, preservando a quantidade original de informações das três bases.

````python

# ============================================================
# 3.3 - VERIFICAÇÃO DE REGISTROS DUPLICADOS
# ============================================================

# Organizar as bases para realizar a verificação
bases_duplicidade = {
    "Pluviométricos": pluviometricos_limpa,
    "Hidrológicos": hidrologicos_limpa,
    "Geotécnicos": geotecnicos_limpa
}

print("=" * 70)
print("STORMWATCH - VERIFICAÇÃO DE REGISTROS DUPLICADOS")
print("=" * 70)

for nome, df in bases_duplicidade.items():

    print(f"\nBASE: {nome.upper()}")

    # Definir os campos utilizados para identificar duplicidades
    colunas_verificacao = ["codEstacao", "datahora"]

    # Nas bases hidrológica e geotécnica, considerar também o sensor
    if nome in ["Hidrológicos", "Geotécnicos"]:
        colunas_verificacao.append("sensor")

    # Conferir se todos os campos necessários existem
    colunas_disponiveis = [
        coluna for coluna in colunas_verificacao
        if coluna in df.columns
    ]

    if len(colunas_disponiveis) != len(colunas_verificacao):
        print("Não foi possível realizar a verificação.")
        print("Colunas necessárias não encontradas.")
        continue

    # Identificar registros duplicados pela combinação dos campos
    duplicados = df.duplicated(
        subset=colunas_verificacao,
        keep=False
    )

    quantidade_duplicados = duplicados.sum()

    print("Campos utilizados:", colunas_verificacao)
    print(f"Registros duplicados encontrados: {quantidade_duplicados:,}")

    if quantidade_duplicados == 0:
        print("Nenhuma duplicidade identificada.")
    else:
        print("Atenção: foram encontradas possíveis duplicidades.")
        display(df.loc[duplicados, colunas_verificacao].head(10))

print("\nVERIFICAÇÃO DE DUPLICIDADES CONCLUÍDA.")
````

<a id="sec-3-4"></a>
## **3.4 Tratamento de valores sentinela nos dados geotécnicos**

Durante a análise da base geotécnica, foi identificado o valor **-1000** em registros de sensores relacionados à umidade do solo.

Esse valor foi identificado como um sentinela, ou seja, um código utilizado para representar uma medição inválida ou indisponível, e não uma medição real de umidade.

Foram identificados **10.038 registros** com esse valor, distribuídos entre sensores de umidade do solo.

O tratamento aplicado foi:

- Preservação do valor original na coluna `valorMedida_original`;
- Identificação dos registros na coluna `status_qualidade`;
- Substituição de -1000 por `NaN` na coluna `valorMedida`, utilizada nas análises.

Essa decisão foi tomada porque manter o valor -1000 nos cálculos poderia distorcer resultados como médias, medianas e desvios padrão.

Os registros não foram excluídos. A substituição por valor ausente permite evitar a utilização de medições inválidas sem perder o histórico original dos dados.



````python

# ============================================================
# 3.4 - TRATAMENTO DE VALORES SENTINELA GEOTÉCNICOS
# ============================================================

print("=" * 70)
print("STORMWATCH - TRATAMENTO DE VALORES SENTINELA")
print("=" * 70)

# Preservar os valores originais antes do tratamento
geotecnicos_limpa["valorMedida_original"] = (
    geotecnicos_limpa["valorMedida"].copy()
)

# Identificar os registros com valor sentinela -1000
valores_sentinela = geotecnicos_limpa["valorMedida"] == -1000

quantidade_sentinela = valores_sentinela.sum()

print(f"\nRegistros com valor -1000 encontrados: {quantidade_sentinela:,}")

# Criar a coluna de qualidade, caso ainda não exista
if "status_qualidade" not in geotecnicos_limpa.columns:
    geotecnicos_limpa["status_qualidade"] = "válido"

# Sinalizar os registros identificados
geotecnicos_limpa.loc[
    valores_sentinela,
    "status_qualidade"
] = "valor sentinela -1000"

# Substituir o sentinela por NaN na coluna analítica
geotecnicos_limpa.loc[
    valores_sentinela,
    "valorMedida"
] = np.nan

# Apresentar os resultados do tratamento
print("\nRESULTADO DO TRATAMENTO:")
print(f"Registros preservados: {len(geotecnicos_limpa):,}")
print(f"Valores sentinela tratados: {quantidade_sentinela:,}")
print(
    "Valores ausentes na coluna valorMedida:",
    geotecnicos_limpa["valorMedida"].isna().sum()
)

print("\nDISTRIBUIÇÃO DOS STATUS DE QUALIDADE:")
print(geotecnicos_limpa["status_qualidade"].value_counts(dropna=False))

print("\nTRATAMENTO DOS VALORES SENTINELA CONCLUÍDO.")
````

<a id="sec-3-5"></a>
## **3.5 Sinalização de valores nos dados hidrológicos**

O tratamento aplicado foi:

- Identificação dos registros na coluna `status_qualidade`;
- Classificação como **"valor 35 - requer investigação"**.

Essa decisão evita excluir informações que podem representar medições legítimas. Os registros permanecem disponíveis, mas devem ser considerados com cautela nas interpretações.

## **3.9. Exportando dados em JSON**

Após a conclusão das etapas de limpeza, validação e preparação das bases pluviométrica, hidrológica e geotécnica, os dados tratados serão exportados para o formato JSON (JavaScript Object Notation).

Essa exportação tem como finalidade integrar os resultados do processamento realizado no notebook Jupyter ao dashboard interativo da plataforma StormWatch. O formato JSON permite que a aplicação web consuma os dados diretamente, utilizando-os para atualizar indicadores, filtros, gráficos e tabelas.

O arquivo gerado será disponibilizado em `data/dashboard/resumo.json`, na raiz do repositório do projeto. Dessa forma, o dashboard poderá acessar uma versão estruturada e padronizada dos dados, sem precisar executar novamente os procedimentos de limpeza e tratamento.

A exportação também facilita a integração entre o ambiente de análise em Python e a interface desenvolvida com HTML, CSS e JavaScript, mantendo o notebook como responsável pelo processamento analítico e o dashboard como responsável pela apresentação interativa dos resultados.

````python

# ============================================================
# 3.9 - EXPORTAÇÃO DOS DADOS PARA JSON
# ============================================================

import json
from pathlib import Path

print("=" * 65)
print("STORMWATCH - EXPORTAÇÃO DOS DADOS PARA JSON")
print("=" * 65)

# ------------------------------------------------------------
# 1. ORGANIZAÇÃO DAS BASES
# ------------------------------------------------------------

# Relaciona cada DataFrame tratado à sua fonte de dados.
bases_exportacao = {
    "chuva": pluviometricos_limpa,
    "hidrologia": hidrologicos_limpa,
    "geotecnia": geotecnicos_limpa
}

# ------------------------------------------------------------
# 2. PREPARAÇÃO DOS REGISTROS
# ------------------------------------------------------------

def preparar_base(df, fonte):

    # Trabalhar com uma cópia para não modificar os DataFrames originais.
    dados = df.copy()

    # Define o sensor da base pluviométrica,
    # que não possui uma coluna específica para essa informação.
    if "sensor" not in dados.columns:
        dados["sensor"] = "chuva"

    # Padroniza os valores utilizados pelo dashboard.
    dados["municipality"] = (
        dados["municipio"]
        .fillna("Não informado")
        .astype(str)
        .str.strip()
    )

    dados["station"] = (
        dados["codEstacao"]
        .fillna("Não informado")
        .astype(str)
        .str.strip()
    )

    dados["station_name"] = (
        dados["nomeEstacao"]
        .fillna("Não informado")
        .astype(str)
        .str.strip()
    )

    dados["sensor"] = (
        dados["sensor"]
        .fillna("Não informado")
        .astype(str)
        .str.strip()
    )

    # Garante que os valores das medições sejam numéricos.
    dados["value"] = pd.to_numeric(
        dados["valorMedida"],
        errors="coerce"
    )

    # Converte a data e mantém somente o dia,
    # pois o dashboard apresenta os dados diariamente.
    dados["date"] = pd.to_datetime(
        dados["datahora"],
        errors="coerce"
    ).dt.strftime("%Y-%m-%d")

    # Remove somente registros sem data válida.
    # Valores de medição ausentes continuam representados
    # nas respectivas agregações.
    dados = dados.dropna(subset=["date"]).copy()

    # Identifica a origem de cada registro.
    dados["source"] = fonte

    return dados


# ------------------------------------------------------------
# 3. AGREGAÇÃO DOS DADOS
# ------------------------------------------------------------

bases_preparadas = []

for fonte, df in bases_exportacao.items():

    dados = preparar_base(df, fonte)

    bases_preparadas.append(dados)

    print(
        f"{fonte.capitalize()}: "
        f"{len(dados):,} registros com data válida"
    )

# Reúne as três fontes em uma única estrutura.
dados_integrados = pd.concat(
    bases_preparadas,
    ignore_index=True
)

# Define os campos utilizados para agrupar os registros.
colunas_agrupamento = [
    "source",
    "date",
    "municipality",
    "station",
    "station_name",
    "sensor"
]

# Calcula as métricas necessárias para o dashboard.
dados_agregados = (
    dados_integrados
    .groupby(
        colunas_agrupamento,
        dropna=False
    )
    .agg(
        count=("value", "count"),
        sum=("value", lambda valores: valores.sum(min_count=1)),
        avg=("value", "mean"),
        min=("value", "min"),
        max=("value", "max")
    )
    .reset_index()
)

# Converte contagens para números inteiros.
dados_agregados["count"] = (
    dados_agregados["count"].astype(int)
)

# Grupos sem nenhuma medição válida recebem soma zero.
# Média, mínimo e máximo permanecem ausentes (null no JSON).
dados_agregados["sum"] = (
    dados_agregados["sum"].fillna(0)
)

# Organiza as colunas na estrutura esperada pelo JavaScript.
dados_agregados = dados_agregados[
    [
        "source",
        "date",
        "municipality",
        "station",
        "station_name",
        "sensor",
        "count",
        "sum",
        "avg",
        "min",
        "max"
    ]
]

# ------------------------------------------------------------
# 4. DEFINIÇÃO DO PERÍODO ANALISADO
# ------------------------------------------------------------

datas_validas = pd.to_datetime(
    dados_integrados["date"],
    errors="coerce"
).dropna()

if datas_validas.empty:
    raise ValueError(
        "Não existem datas válidas para gerar o arquivo JSON."
    )

periodo = {
    "start": datas_validas.min().strftime("%Y-%m-%d"),
    "end": datas_validas.max().strftime("%Y-%m-%d")
}

# ------------------------------------------------------------
# 5. METADADOS DAS FONTES
# ------------------------------------------------------------

source_metadata = {}

for fonte, df in bases_exportacao.items():

    dados_fonte = dados_integrados[
        dados_integrados["source"] == fonte
    ]

    source_metadata[fonte] = {
        "total_records": int(len(df)),
        "valid_date_records": int(len(dados_fonte)),
        "stations": int(
            dados_fonte["station"].nunique()
        ),
        "municipalities": int(
            dados_fonte["municipality"].nunique()
        )
    }

# ------------------------------------------------------------
# 6. MONTAGEM DA ESTRUTURA JSON
# ------------------------------------------------------------

# O to_json converte valores ausentes para null,
# evitando a geração de NaN, que não é um valor JSON válido.
registros_json = json.loads(
    dados_agregados.to_json(
        orient="records",
        force_ascii=False,
        double_precision=10
    )
)

resumo = {
    "period": periodo,

    "source_metadata": source_metadata,

    "total_records": int(
        sum(len(df) for df in bases_exportacao.values())
    ),

    "aggregated_records": len(registros_json),

    "records": registros_json
}

# ------------------------------------------------------------
# 7. DEFINIÇÃO DO CAMINHO DE EXPORTAÇÃO
# ------------------------------------------------------------

# Procura a raiz do projeto a partir do diretório atual.
diretorio_atual = Path.cwd()
raiz_projeto = None

for candidato in [diretorio_atual, *diretorio_atual.parents]:

    if (
        (candidato / ".git").exists()
        or (candidato / "global-solution").is_dir()
    ):
        raiz_projeto = candidato
        break

# Caso o notebook esteja sendo executado diretamente
# dentro da pasta global-solution.
if raiz_projeto is None and diretorio_atual.name == "global-solution":
    raiz_projeto = diretorio_atual.parent

# Se não encontrar a estrutura do repositório,
# utiliza o diretório atual como destino.
if raiz_projeto is None:
    raiz_projeto = diretorio_atual

# Cria a pasta de destino caso ela ainda não exista.
pasta_destino = raiz_projeto / "data" / "dashboard"
pasta_destino.mkdir(parents=True, exist_ok=True)

caminho_json = pasta_destino / "resumo.json"

# ------------------------------------------------------------
# 8. GRAVAÇÃO DO ARQUIVO
# ------------------------------------------------------------

with open(
    caminho_json,
    "w",
    encoding="utf-8"
) as arquivo:

    json.dump(
        resumo,
        arquivo,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )

# ------------------------------------------------------------
# 9. VALIDAÇÃO DA EXPORTAÇÃO
# ------------------------------------------------------------

# Reabre o arquivo para verificar se o JSON foi gravado corretamente.
with open(
    caminho_json,
    "r",
    encoding="utf-8"
) as arquivo:

    json_validado = json.load(arquivo)

print("\n" + "=" * 65)
print("EXPORTAÇÃO CONCLUÍDA COM SUCESSO!")
print("=" * 65)

print(f"\nArquivo gerado: {caminho_json}")
print(f"Período: {periodo['start']} até {periodo['end']}")

print(
    f"Total de registros originais: "
    f"{json_validado['total_records']:,}"
)

print(
    f"Total de registros agregados: "
    f"{json_validado['aggregated_records']:,}"
)

print(f"Tamanho do arquivo: {caminho_json.stat().st_size / 1024:.2f} KB")

print("\nESTRUTURA DO JSON VALIDADA.")
print("O arquivo está pronto para ser consumido pelo dashboard StormWatch.")
````

# **5. Técnicas adicionais**

## **5.1 Integração de dados pluviométricos e hidrológicos**

Nesta etapa, foi realizada a integração entre as bases pluviométrica e hidrológica, utilizando a biblioteca Pandas.

Inicialmente, os registros hidrológicos foram separados em duas categorias: sensores de chuva e sensores de nível da água.

Em seguida, foram realizados dois cruzamentos de dados:

- **Integração pluviométrica e chuva hidrológica:** relaciona as medições das duas bases que possuem o mesmo código de estação e a mesma data e horário.
- **Integração pluviométrica e nível hidrológico:** relaciona os registros pluviométricos com as medições de nível da água que possuem correspondência na mesma estação e no mesmo instante.

Para realizar os cruzamentos, foi utilizado o método `merge()` do Pandas, com os campos `codEstacao` e `datahora` como critérios de correspondência e o parâmetro `how="inner"`, que mantém somente os registros encontrados nas duas bases.

Após a integração, foram calculadas a quantidade de registros correspondentes, a quantidade de estações identificadas e, no cruzamento com os dados de nível, a quantidade de registros em que os municípios das duas bases apresentam o mesmo nome.

Essa técnica permite verificar a possibilidade de relacionar diferentes fontes de informações ambientais e constitui uma etapa importante para a análise conjunta dos dados utilizados pelo StormWatch.

É importante destacar que a correspondência entre registros não comprova uma relação de causa e efeito entre precipitação e nível da água. Ela demonstra apenas que existem medições associadas à mesma estação e ao mesmo horário.

````python
# ================================================================
# INTEGRAÇÃO DOS DADOS PLUVIOMÉTRICOS E HIDROLÓGICOS
# ================================================================

# SEPARAÇÃO DOS REGISTROS HIDROLÓGICOS POR TIPO DE SENSOR

# Cria uma base contendo somente os registros dos sensores de chuva.
# O comando .copy() cria uma cópia independente dos dados selecionados.
hidro_chuva = hidrologicos_limpa[
    hidrologicos_limpa["sensor"] == "chuva"
].copy()

# Cria outra base contendo somente os registros dos sensores de nível da água.
hidro_nivel = hidrologicos_limpa[
    hidrologicos_limpa["sensor"] == "nível"
].copy()


# ================================================================
# INTEGRAÇÃO DOS DADOS PLUVIOMÉTRICOS COM OS SENSORES DE CHUVA
# ================================================================

# Utiliza o merge() do Pandas para relacionar os dados pluviométricos
# com os registros de chuva da base hidrológica.
#
# A correspondência é realizada por dois campos:
# codEstacao: identifica a estação de monitoramento.
# datahora: identifica a data e o horário da medição.
#
# how="inner" mantém somente os registros que possuem correspondência
# nas duas bases.
#
# suffixes diferencia os nomes das colunas que existem nas duas bases,
# utilizando "_pluvio" para a base pluviométrica e "_hidro" para a hidrológica.
integracao_chuva = pd.merge(
    pluviometricos_limpa,
    hidro_chuva,
    on=["codEstacao", "datahora"],
    how="inner",
    suffixes=("_pluvio", "_hidro")
)


# ================================================================
# INTEGRAÇÃO DOS DADOS PLUVIOMÉTRICOS COM OS SENSORES DE NÍVEL
# ================================================================

# Realiza outro cruzamento, agora relacionando os dados pluviométricos
# com os registros dos sensores de nível da água.
#
# São utilizados os mesmos critérios de correspondência:
# código da estação e data e horário da medição.
#
# O resultado contém somente os registros encontrados nas duas bases.
integracao_nivel = pd.merge(
    pluviometricos_limpa,
    hidro_nivel,
    on=["codEstacao", "datahora"],
    how="inner",
    suffixes=("_pluvio", "_hidro")
)


# ================================================================
# APRESENTAÇÃO DOS RESULTADOS DA INTEGRAÇÃO
# ================================================================

# Exibe o título da análise
print("INTEGRAÇÃO PLUVIOMÉTRICA E HIDROLÓGICA")


# RESULTADOS DA INTEGRAÇÃO ENTRE OS REGISTROS DE CHUVA

print("\nComparação entre chuva:")

# Conta quantos registros foram encontrados nas duas bases
print("Registros correspondentes:", len(integracao_chuva))

# Conta quantas estações diferentes participaram da integração
# O nunique() considera cada código de estação apenas uma vez
print(
    "Estações correspondentes:",
    integracao_chuva["codEstacao"].nunique()
)


# RESULTADOS DA INTEGRAÇÃO ENTRE CHUVA E NÍVEL DA ÁGUA

print("\nComparação entre chuva e nível:")

# Conta quantos registros pluviométricos e hidrológicos
# apresentaram correspondência na mesma estação e horário
print("Registros correspondentes:", len(integracao_nivel))

# Conta quantas estações diferentes participaram dessa integração
print(
    "Estações correspondentes:",
    integracao_nivel["codEstacao"].nunique()
)


# COMPARAÇÃO DOS MUNICÍPIOS NA INTEGRAÇÃO DE NÍVEL

print("\nMunicípios correspondentes na integração de nível:")

# Compara os municípios registrados nas duas bases.
# A condição retorna True quando os nomes dos municípios são iguais.
# O sum() conta quantas correspondências foram encontradas.
print(
    (integracao_nivel["municipio_pluvio"] ==
     integracao_nivel["municipio_hidro"]).sum()
)
````

## **5.2 Correlação entre variáveis**

A segunda técnica adicional aplicada ao projeto foi a análise de correlação, utilizando os métodos de Pearson e Spearman. Essa técnica permite investigar a existência, a direção e a intensidade de associações estatísticas entre duas variáveis.

Foram realizadas duas comparações:

- **Chuva pluviométrica x chuva hidrológica:** busca identificar se as medições de chuva das duas bases apresentam associação.
- **Chuva pluviométrica x nível hidrológico:** busca investigar se as medições de chuva apresentam associação com as variações do nível hidrológico.

O método de Pearson avalia relações lineares entre as variáveis, enquanto Spearman avalia relações monotônicas com base na ordenação dos valores.

Os resultados variam entre -1 e +1. Valores positivos indicam associação positiva, valores negativos indicam associação negativa e valores próximos de zero indicam uma associação linear ou monotônica fraca, conforme o método utilizado.

A aplicação dessa técnica complementa a integração das bases de dados, permitindo investigar possíveis relações entre os fenômenos ambientais monitorados pelo StormWatch.

**Interpretação dos resultados:** os valores de correlação obtidos nas duas comparações devem ser analisados separadamente. A correlação indica associação estatística, mas não comprova causalidade. Além disso, a correspondência temporal utilizada considera registros da mesma estação e horário, não avaliando possíveis efeitos com defasagem temporal.



````python
# ================================================================
# CÁLCULO DE CORRELAÇÕES ENTRE OS DADOS INTEGRADOS
# ================================================================

# Cria uma função para calcular a correlação entre duas variáveis.
# A função pode ser reutilizada em diferentes conjuntos de dados.
#
# Parâmetros:
# df: conjunto de dados que será analisado;
# coluna_a: nome da primeira variável;
# coluna_b: nome da segunda variável;
# nome: identificação da análise apresentada nos resultados.
def analisar_correlacao(df, coluna_a, coluna_b, nome):

    # Seleciona somente as duas variáveis que serão comparadas.
    # O dropna() remove os registros que possuem valores ausentes
    # em qualquer uma das duas colunas selecionadas.
    dados = df[[coluna_a, coluna_b]].dropna()

    # Apresenta o nome da análise e a quantidade de pares válidos.
    print(f"\n--- {nome} ---")
    print("Quantidade de pares válidos:", len(dados))

    # Verifica se existem pelo menos dois pares de valores válidos.
    # Caso contrário, interrompe a análise para evitar um cálculo inválido.
    if len(dados) < 2:
        print("Dados insuficientes para calcular a correlação.")
        return

    # Verifica se alguma das variáveis possui apenas um valor diferente.
    # Uma variável constante não permite calcular a correlação.
    if dados[coluna_a].nunique() < 2 or dados[coluna_b].nunique() < 2:
        print("Não é possível calcular: uma das variáveis é constante.")
        return

    # Calcula a correlação de Pearson.
    # Esse método mede a intensidade e a direção da relação linear
    # entre as duas variáveis.
    # O resultado varia de -1 a +1.
    print("Correlação de Pearson:", dados[coluna_a].corr(
        dados[coluna_b], method="pearson"
    ))

    # Calcula a correlação de Spearman.
    # Esse método considera a posição relativa (ordem) dos valores
    # e permite investigar relações monotônicas entre as variáveis.
    # O resultado também varia de -1 a +1.
    print("Correlação de Spearman:", dados[coluna_a].corr(
        dados[coluna_b], method="spearman"
    ))


# ================================================================
# 1. CORRELAÇÃO ENTRE CHUVA PLUVIOMÉTRICA E CHUVA HIDROLÓGICA
# ================================================================

# Utiliza a base integrada de chuva para investigar se as medições
# pluviométricas e hidrológicas apresentam associação estatística.
#
# São comparadas as colunas:
# valorMedida_pluvio: medição da base pluviométrica;
# valorMedida_hidro: medição de chuva da base hidrológica.
analisar_correlacao(
    integracao_chuva,
    "valorMedida_pluvio",
    "valorMedida_hidro",
    "Chuva pluviométrica x chuva hidrológica"
)


# ================================================================
# 2. CORRELAÇÃO ENTRE CHUVA PLUVIOMÉTRICA E NÍVEL HIDROLÓGICO
# ================================================================

# Utiliza a base integrada de nível para investigar se existe
# associação estatística entre as medições de chuva e o nível da água.
#
# São comparadas as colunas:
# valorMedida_pluvio: medição da base pluviométrica;
# valorMedida_hidro: medição de nível da base hidrológica.
analisar_correlacao(
    integracao_nivel,
    "valorMedida_pluvio",
    "valorMedida_hidro",
    "Chuva pluviométrica x nível hidrológico"
)
````

## **5.3 Probabilidade condicional: chuva e nível hidrológico**

A terceira técnica adicional aplicada ao projeto StormWatch foi a análise de probabilidade, utilizando o conceito de probabilidade condicional. Essa técnica foi escolhida para investigar a relação entre a ocorrência de chuva e a presença de níveis hidrológicos relativamente elevados, permitindo avaliar como esses fenômenos se apresentam conjuntamente nos registros ambientais.

A análise utiliza a base integrada de dados pluviométricos e hidrológicos, considerando registros correspondentes à mesma estação e ao mesmo horário. Foram estabelecidas duas condições para identificar os eventos analisados:

- **Ocorrência de chuva:** registros em que o valor da medição pluviométrica é maior que zero.
- **Nível hidrológico elevado:** registros em que o valor da medição hidrológica é igual ou superior ao percentil 75 da distribuição analisada.

O percentil 75 foi utilizado como critério estatístico para identificar os valores relativamente mais elevados do conjunto de dados. Dessa forma, a classificação não representa um limite oficial de alerta ou risco de inundação, mas permite comparar os registros com base em sua própria distribuição.

### **Probabilidades calculadas**

Foram calculadas quatro probabilidades para investigar o comportamento conjunto das variáveis:

- **P(chuva):** representa a proporção de registros em que houve ocorrência de chuva.
- **P(nível elevado):** representa a proporção de registros classificados como nível hidrológico elevado.
- **P(chuva | nível elevado):** representa a probabilidade de ocorrência de chuva considerando apenas os registros em que o nível hidrológico estava elevado.
- **P(nível elevado | chuva):** representa a probabilidade de ocorrência de nível hidrológico elevado considerando apenas os registros em que houve chuva.

Essas medidas permitem observar diferentes perspectivas da associação entre chuva e nível hidrológico, contribuindo para compreender como as duas variáveis se comportam nos dados integrados.

### **Comparação entre cenários**

Para avaliar a influência dos registros hidrológicos sinalizados anteriormente, a análise foi executada em dois cenários:

1. **Todos os registros:** considera a totalidade dos dados integrados, incluindo os registros com nível hidrológico igual a 35.
2. **Sem os registros de nível 35:** exclui temporariamente esses registros para verificar como sua presença influencia as probabilidades calculadas.

Essa comparação permite identificar possíveis alterações nos resultados e avaliar a sensibilidade da análise aos registros que ainda precisam de validação.

### **Interpretação dos resultados**

Os resultados devem ser interpretados comparando as probabilidades obtidas nos dois cenários. A diferença entre P(chuva | nível elevado) e P(nível elevado | chuva) é importante, pois cada medida considera uma condição diferente.

Caso a probabilidade de nível elevado seja maior entre os registros com chuva do que na totalidade dos registros, isso poderá indicar uma associação entre a ocorrência de chuva e níveis hidrológicos elevados. Entretanto, essa associação não comprova que a chuva tenha causado a elevação do nível, pois outros fatores também podem influenciar o comportamento hidrológico.

### **Contribuição para o StormWatch**

A aplicação da probabilidade condicional amplia a capacidade analítica do StormWatch, permitindo investigar possíveis associações entre diferentes variáveis ambientais. Os indicadores calculados podem contribuir para a compreensão dos fenômenos monitorados e apoiar futuras análises relacionadas ao acompanhamento de condições hidrológicas.

É importante destacar que o critério de nível elevado foi definido estatisticamente pelo percentil 75, e não por parâmetros oficiais de alerta. Além disso, a análise considera registros no mesmo horário, não contemplando possíveis atrasos entre a ocorrência da chuva e a resposta do nível hidrológico.

````python
# ================================================================
# PROBABILIDADE CONDICIONAL: CHUVA E NÍVEL HIDROLÓGICO
# ================================================================

# Cria uma função para calcular diferentes probabilidades
# relacionadas à ocorrência de chuva e ao nível hidrológico.
#
# Parâmetros:
# df: conjunto de dados integrado que será analisado;
# descricao: identificação do cenário apresentado nos resultados.
def calcular_probabilidades(df, descricao):

    # Seleciona somente as colunas de chuva pluviométrica
    # e nível hidrológico.
    #
    # O dropna() remove registros que possuem valores ausentes
    # nessas duas variáveis.
    # O copy() cria uma cópia para evitar alterações na base original.
    dados = df[
        ["valorMedida_pluvio", "valorMedida_hidro"]
    ].dropna().copy()

    # Define o limite de nível hidrológico elevado utilizando
    # o percentil 75 (terceiro quartil).
    # Isso significa que os valores iguais ou superiores a esse
    # limite representam aproximadamente os 25% maiores valores
    # observados na distribuição.
    limite_nivel = dados["valorMedida_hidro"].quantile(0.75)

    # Cria uma variável chamada "chuva".
    # Recebe True quando existe registro de chuva (valor maior que zero)
    # e False quando o valor é zero.
    dados["chuva"] = dados["valorMedida_pluvio"] > 0

    # Cria uma variável chamada "nivel_elevado".
    # Recebe True quando o nível hidrológico é igual ou superior
    # ao limite definido pelo percentil 75.
    dados["nivel_elevado"] = dados["valorMedida_hidro"] >= limite_nivel

    # ============================================================
    # CÁLCULO DAS PROBABILIDADES
    # ============================================================

    # Calcula a probabilidade de ocorrência de chuva.
    # A média de uma coluna booleana considera True como 1 e False como 0.
    p_chuva = dados["chuva"].mean()

    # Calcula a probabilidade de ocorrência de nível hidrológico elevado.
    p_nivel_elevado = dados["nivel_elevado"].mean()

    # Calcula a probabilidade de haver chuva dado que o nível
    # hidrológico está elevado.
    # Considera somente os registros classificados como nível elevado.
    p_chuva_dado_nivel = (
        dados.loc[dados["nivel_elevado"], "chuva"].mean()
    )

    # Calcula a probabilidade de o nível hidrológico estar elevado
    # dado que existe registro de chuva.
    # Considera somente os registros em que houve chuva.
    p_nivel_dado_chuva = (
        dados.loc[dados["chuva"], "nivel_elevado"].mean()
    )

    # ============================================================
    # APRESENTAÇÃO DOS RESULTADOS
    # ============================================================

    # Exibe a identificação do cenário analisado.
    print(f"\n--- {descricao} ---")

    # Informa a quantidade de registros válidos utilizados.
    print("Quantidade de registros:", len(dados))

    # Apresenta o limite de nível hidrológico definido pelo percentil 75.
    print("Percentil 75 do nível:", limite_nivel)

    # Apresenta a probabilidade de ocorrência de chuva.
    # O round() limita a apresentação a quatro casas decimais.
    print(
        "P(chuva):",
        round(p_chuva, 4)
    )

    # Apresenta a probabilidade de ocorrência de nível elevado.
    print(
        "P(nível elevado):",
        round(p_nivel_elevado, 4)
    )

    # Apresenta a probabilidade de chuva quando o nível hidrológico
    # está elevado.
    print(
        "P(chuva | nível elevado):",
        round(p_chuva_dado_nivel, 4)
    )

    # Apresenta a probabilidade de nível elevado quando existe
    # registro de chuva.
    print(
        "P(nível elevado | chuva):",
        round(p_nivel_dado_chuva, 4)
    )

# ================================================================
# 1. ANÁLISE INCLUINDO TODOS OS REGISTROS
# ================================================================

# Executa a função considerando todos os registros da base integrada,
# inclusive aqueles que apresentam nível hidrológico igual a 35.
calcular_probabilidades(
    integracao_nivel,
    "Análise com todos os registros"
)

# ================================================================
# 2. ANÁLISE SEM OS REGISTROS SINALIZADOS COM NÍVEL 35
# ================================================================

# Cria uma nova base excluindo os registros cujo nível hidrológico
# é exatamente igual a 35.
#
# Essa exclusão é feita somente para comparar os resultados
# com o cenário anterior, sem modificar a base original.
integracao_nivel_sem_35 = integracao_nivel[
    integracao_nivel["valorMedida_hidro"] != 35
].copy()

# Executa novamente os cálculos utilizando a base sem os registros
# de nível 35, permitindo observar possíveis diferenças
# nas probabilidades calculadas.
calcular_probabilidades(
    integracao_nivel_sem_35,
    "Análise sem os registros de nível 35"
)
````

Entretanto, as probabilidades são condicionadas aos registros disponíveis e aos critérios adotados, não representando previsões oficiais de enchentes ou deslizamentos.

<a id="sec-8-1"></a>
# **8. Limitações e uso responsável dos dados**

A análise do StormWatch apresenta limitações relacionadas à qualidade, à origem, à abrangência e ao período dos dados utilizados. Esses fatores precisam ser considerados na interpretação dos resultados e no desenvolvimento de funcionalidades de monitoramento ambiental.

## **8.1 Limitações das bases de dados**

As bases utilizadas abrangem o período de 1º a 31 de janeiro de 2026. Portanto, representam apenas um mês de observações e não permitem identificar, isoladamente, padrões sazonais, tendências anuais ou comportamentos de longo prazo.

A abrangência geográfica também é diferente entre as bases: os dados pluviométricos contemplam 193 municípios, os hidrológicos 20 municípios e os geotécnicos 8 municípios. Essa diferença limita as comparações entre regiões, pois nem todos os municípios possuem a mesma disponibilidade de medições.

Além disso, a origem institucional, os endereços oficiais de disponibilização, as licenças de uso e as unidades de algumas variáveis ainda precisam ser confirmados. Dessa forma, os valores não devem ser interpretados como medidas físicas plenamente validadas sem a confirmação dos metadados das fontes.

<a id="sec-8-2"></a>
## **8.2 Qualidade e tratamento dos dados**

Durante a preparação das bases, foram identificadas situações que exigiram tratamento específico:

- Na base geotécnica, foram encontrados 10.038 registros com o valor sentinela `-1000`, tratado como dado ausente nas análises, preservando-se o valor original.
- Na base hidrológica, 2.627 registros com valor de nível igual a `35` foram mantidos e sinalizados para investigação, sem remoção automática.
- Dois vínculos entre estações geotécnicas e pluviométricas foram classificados como aproximações espaciais.
- A integração temporal foi realizada por correspondência de estação e horário, sem interpolação de valores ausentes.

Esses procedimentos preservam a rastreabilidade dos dados, mas não eliminam todas as incertezas. Em especial, os valores sinalizados precisam ser verificados com as fontes responsáveis antes de serem utilizados em decisões operacionais.

<a id="sec-8-3"></a>
## **8.3 Representatividade e possíveis vieses**

A distribuição das estações não é uniforme entre os municípios. Algumas regiões possuem maior quantidade de registros e sensores, enquanto outras apresentam cobertura limitada ou inexistente.

Isso pode gerar um viés de cobertura: municípios com mais estações tendem a aparecer com maior frequência nas análises, enquanto regiões sem monitoramento podem ficar sub-representadas.

Também é importante considerar que a quantidade de medições não representa diretamente a quantidade de habitantes afetados. Os dados descrevem os locais monitorados pelos sensores, e não toda a população de cada município.

Assim, grupos que vivem em regiões rurais, afastadas ou com pouca infraestrutura de monitoramento podem não estar adequadamente representados nas bases. A ausência de registros em determinada região não significa ausência de risco ambiental.

<a id="sec-8-4"></a>
## **8.4 Privacidade e uso responsável (LGPD)**

O projeto trabalha principalmente com informações ambientais, como localização de estações, datas, horários e valores de sensores. Entretanto, caso sejam incorporados dados pessoais ou informações que permitam identificar indivíduos, será necessário aplicar os cuidados previstos na Lei Geral de Proteção de Dados Pessoais (LGPD).

Entre as medidas de proteção estão:

- Evitar a coleta de dados pessoais que não sejam necessários ao funcionamento da plataforma.
- Remover nomes, documentos, contatos e outros identificadores pessoais antes da disponibilização dos dados.
- Aplicar anonimização quando houver informações pessoais que possam identificar indivíduos.
- Restringir o acesso a informações que não precisem ser públicas.
- Utilizar os dados somente para as finalidades definidas para o projeto.

A localização de uma estação de monitoramento não deve ser confundida automaticamente com a localização de uma pessoa. Ainda assim, eventuais dados adicionais associados a indivíduos precisam ser avaliados quanto ao risco de identificação.

<a id="sec-8-5"></a>
## **8.5 Cuidados na interpretação e utilização dos resultados**

Os resultados apresentados pelo StormWatch devem ser utilizados como apoio à visualização, integração e análise de informações ambientais. As médias, correlações e contagens identificadas não representam, por si só, comprovação de risco ou previsão de desastres.

A ausência de unidades confirmadas, de limites oficiais de alerta e de validação independente de algumas medições impede que os resultados sejam tratados como critérios definitivos para emissão de alertas.

Portanto, antes de implementar classificações de risco ou alertas automáticos, será necessário validar as fontes, confirmar as unidades de medição, consultar critérios técnicos oficiais e avaliar os resultados com especialistas da área.

Dessa forma, o StormWatch busca promover o uso responsável dos dados ambientais, reconhecendo suas limitações e evitando interpretações que possam gerar conclusões equivocadas ou decisões inadequadas.
